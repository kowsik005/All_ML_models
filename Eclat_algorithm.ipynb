```python
# ==========================================================
# ECLAT ALGORITHM - GOOGLE COLAB
# ==========================================================

# Step 1: Import libraries
from itertools import combinations


# ==========================================================
# Step 2: Create transaction dataset
# ==========================================================

transactions = [
    ['Milk', 'Bread', 'Butter'],
    ['Bread', 'Butter'],
    ['Milk', 'Bread'],
    ['Milk', 'Eggs'],
    ['Bread', 'Eggs'],
    ['Milk', 'Bread', 'Eggs'],
    ['Milk', 'Bread', 'Butter'],
    ['Bread', 'Butter', 'Eggs'],
    ['Milk', 'Butter'],
    ['Milk', 'Bread', 'Butter', 'Eggs']
]

print("Transactions:")
for i, transaction in enumerate(transactions, start=1):
    print(f"T{i}: {transaction}")


# ==========================================================
# Step 3: Create TID sets
# ==========================================================

# TID = Transaction ID
# Store the transaction IDs in which each item appears.

tidsets = {}

for tid, transaction in enumerate(transactions, start=1):

    for item in transaction:

        if item not in tidsets:
            tidsets[item] = set()

        tidsets[item].add(tid)


print("\n==========================================")
print("TID SETS")
print("==========================================")

for item, tids in tidsets.items():
    print(f"{item}: {sorted(tids)}")


# ==========================================================
# Step 4: Set minimum support
# ==========================================================

min_support = 0.3

min_count = int(
    min_support * len(transactions)
)

print("\nMinimum Support:", min_support)
print("Minimum Support Count:", min_count)


# ==========================================================
# Step 5: Find frequent 1-itemsets
# ==========================================================

frequent_itemsets = {}

for item, tids in tidsets.items():

    if len(tids) >= min_count:

        itemset = frozenset([item])

        frequent_itemsets[itemset] = tids


# ==========================================================
# Step 6: ECLAT function
# ==========================================================

def eclat(
    current_itemsets,
    min_count,
    all_frequent_itemsets
):

    items = list(current_itemsets.keys())

    for i in range(len(items)):

        item_a = items[i]
        tidset_a = current_itemsets[item_a]

        for j in range(i + 1, len(items)):

            item_b = items[j]
            tidset_b = current_itemsets[item_b]

            # Intersect TID sets
            new_tidset = tidset_a.intersection(tidset_b)

            # Check minimum support
            if len(new_tidset) >= min_count:

                new_itemset = item_a.union(item_b)

                # Avoid duplicates
                if new_itemset not in all_frequent_itemsets:

                    all_frequent_itemsets[
                        new_itemset
                    ] = new_tidset

                    # Continue expanding
                    new_candidates = {
                        new_itemset: new_tidset
                    }

                    eclat(
                        new_candidates,
                        min_count,
                        all_frequent_itemsets
                    )


# ==========================================================
# Step 7: Run ECLAT
# ==========================================================

eclat(
    frequent_itemsets,
    min_count,
    frequent_itemsets
)


# ==========================================================
# Step 8: Display frequent itemsets
# ==========================================================

print("\n==========================================")
print("FREQUENT ITEMSETS")
print("==========================================")

results = []

for itemset, tids in frequent_itemsets.items():

    support_count = len(tids)

    support = (
        support_count /
        len(transactions)
    )

    results.append(
        (
            tuple(sorted(itemset)),
            support_count,
            support
        )
    )


# Sort by itemset size and name
results.sort(
    key=lambda x: (len(x[0]), x[0])
)


for itemset, count, support in results:

    print(
        f"{itemset} "
        f"→ Support Count: {count}, "
        f"Support: {support:.2f}"
    )
```
