```python
# ============================================
# SVM CLASSIFICATION - BANK MARKETING DATASET
# ============================================

import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.svm import SVC
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix
)

# ============================================
# 1. Load Dataset
# ============================================

df = pd.read_csv("bank-additional-full-selected-columns.csv")

print("Dataset Shape:", df.shape)
print("\nColumns:")
print(df.columns)

# ============================================
# 2. Display Dataset Information
# ============================================

print("\nFirst 5 Rows:")
print(df.head())

print("\nMissing Values:")
print(df.isnull().sum())

# ============================================
# 3. Remove Empty Rows
# ============================================

df = df.dropna(how="all")

print("\nShape after removing empty rows:", df.shape)

# ============================================
# 4. Define Features and Target
# ============================================

# Change 'y' if your target column has another name
target_column = "y"

if target_column not in df.columns:
    print("\nERROR: Target column 'y' was not found.")
    print("Available columns:", df.columns.tolist())
else:

    X = df.drop(columns=[target_column])
    y = df[target_column]

    # ========================================
    # 5. Convert Target to 0 and 1
    # ========================================

    if y.dtype == "object":
        y = y.map({
            "no": 0,
            "yes": 1
        })

    # Remove rows where target is missing
    valid_rows = y.notna()
    X = X[valid_rows]
    y = y[valid_rows]

    # ========================================
    # 6. Identify Column Types
    # ========================================

    numerical_columns = X.select_dtypes(
        include=["int64", "float64"]
    ).columns.tolist()

    categorical_columns = X.select_dtypes(
        include=["object"]
    ).columns.tolist()

    print("\nNumerical Columns:")
    print(numerical_columns)

    print("\nCategorical Columns:")
    print(categorical_columns)

    # ========================================
    # 7. Preprocessing
    # ========================================

    preprocessor = ColumnTransformer(
        transformers=[
            (
                "numeric",
                StandardScaler(),
                numerical_columns
            ),
            (
                "categorical",
                OneHotEncoder(
                    handle_unknown="ignore"
                ),
                categorical_columns
            )
        ]
    )

    # ========================================
    # 8. Create SVM Model
    # ========================================

    svm_model = Pipeline(
        steps=[
            ("preprocessor", preprocessor),

            ("svm", SVC(
                kernel="rbf",
                C=1.0,
                gamma="scale"
            ))
        ]
    )

    # ========================================
    # 9. Train-Test Split
    # ========================================

    X_train, X_test, y_train, y_test = train_test_split(
        X,
        y,
        test_size=0.20,
        random_state=42,
        stratify=y
    )

    print("\nTraining Data:", X_train.shape)
    print("Testing Data:", X_test.shape)

    # ========================================
    # 10. Train SVM
    # ========================================

    svm_model.fit(X_train, y_train)

    print("\nSVM Model Training Completed!")

    # ========================================
    # 11. Prediction
    # ========================================

    y_pred = svm_model.predict(X_test)

    # ========================================
    # 12. Accuracy
    # ========================================

    accuracy = accuracy_score(y_test, y_pred)

    print("\n================================")
    print("SVM MODEL RESULTS")
    print("================================")

    print("\nAccuracy:", accuracy)
    print("Accuracy (%):", accuracy * 100)

    # ========================================
    # 13. Classification Report
    # ========================================

    print("\nClassification Report:")
    print(classification_report(y_test, y_pred))

    # ========================================
    # 14. Confusion Matrix
    # ========================================

    print("\nConfusion Matrix:")
    print(confusion_matrix(y_test, y_pred))
```
