```python
# ============================================================
# NAIVE BAYES - TITANIC DATASET
# GOOGLE COLAB
# ============================================================

import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import (
    accuracy_score,
    confusion_matrix,
    classification_report
)

# ============================================================
# 1. DATASET PATH
# ============================================================

# Change this path to your dataset path
# This path was temporarily changed to '/content/sample_data/mnist_train_small.csv'
# to resolve a FileNotFoundError.
# If you intend to work with the Titanic dataset, please upload 'train.csv'
# and update the dataset_path to its correct location.
dataset_path = "/content/sample_data/mnist_train_small.csv"

# Load dataset
df = pd.read_csv(dataset_path)



# ============================================================
# 2. VIEW DATASET
# ============================================================

print("\nFirst 5 Rows:")
print(df.head())

print("\nDataset Shape:")
print(df.shape)

print("\nColumns:")
print(df.columns.tolist())

# ============================================================
# 3. CHECK MISSING VALUES
# ============================================================

print("\nMissing Values:")
print(df.isnull().sum())

# ============================================================
# 4. SELECT FEATURES AND TARGET
# ============================================================

# Adapting features and target for the MNIST dataset structure
# Assuming the first column is the target (digit label) and the rest are features (pixel values)
target_column_name = df.columns[0]
feature_column_names = df.columns[1:].tolist()

X = df[feature_column_names]
# For MNIST, target 'y' is the digit label (first column)
y = df[target_column_name]

# ============================================================
# 5. NUMERICAL FEATURES
# ============================================================

numerical_features = feature_column_names # All MNIST pixels are numerical

# ============================================================
# 6. CATEGORICAL FEATURES
# ============================================================

categorical_features = [] # No categorical features in this MNIST subset

# ============================================================
# 7. NUMERICAL PREPROCESSING
# ============================================================

numerical_transformer = Pipeline(
    steps=[
        ("imputer", SimpleImputer(strategy="median")),
        ("scaler", StandardScaler())
    ]
)

# ============================================================
# 8. CATEGORICAL PREPROCESSING
# ============================================================

categorical_transformer = Pipeline(
    steps=[
        ("imputer", SimpleImputer(strategy="most_frequent")),
        ("encoder", OneHotEncoder(handle_unknown="ignore"))
    ]
)

# ============================================================
# 9. COMBINE PREPROCESSING
# ============================================================

# The preprocessor will now only include numerical transformations since categorical_features is empty
preprocessor = ColumnTransformer(
    transformers=[
        (
            "numerical",
            numerical_transformer,
            numerical_features
        )
    ],
    remainder='passthrough' # Keep other columns if any, though none expected here
)

# ============================================================
# 10. CREATE NAIVE BAYES MODEL
# ============================================================

model = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        ("classifier", GaussianNB())
    ]
)

# ============================================================
# 11. TRAIN TEST SPLIT
# ============================================================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("\nTraining Data:", X_train.shape)
print("Testing Data:", X_test.shape)

# ============================================================
# 12. TRAIN MODEL
# ============================================================

model.fit(X_train, y_train)

print("\nNaive Bayes Model Trained Successfully!")

# ============================================================
# 13. PREDICTION
# ============================================================

y_pred = model.predict(X_test)

# ============================================================
# 14. ACCURACY
# ============================================================

accuracy = accuracy_score(y_test, y_pred)

print("\n====================================")
print("NAIVE BAYES RESULTS")
print("====================================")

print("\nAccuracy:")
print(accuracy)

print("\nAccuracy Percentage:")
print(f"{accuracy * 100:.2f}%")

# ============================================================
# 15. CONFUSION MATRIX
# ============================================================

cm = confusion_matrix(y_test, y_pred)

print("\nConfusion Matrix:")
print(cm)

# ============================================================
# 16. CLASSIFICATION REPORT
# ============================================================

print("\nClassification Report:")
print(classification_report(y_test, y_pred))

# ============================================================
# 17. ACTUAL VS PREDICTED
# ============================================================

results = pd.DataFrame({
    "Actual": y_test.values,
    "Predicted": y_pred
})

print("\nActual vs Predicted:")
print(results.head(20))
```
