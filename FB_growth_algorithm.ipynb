```python
# ==========================================================
# FP-GROWTH ALGORITHM - GOOGLE COLAB
# ==========================================================

# Step 1: Install mlxtend
!pip install mlxtend -q


# Step 2: Import libraries
import pandas as pd

from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import fpgrowth
from mlxtend.frequent_patterns import association_rules


# ==========================================================
# Step 3: Create transaction dataset
# ==========================================================

transactions = [
    ['Milk', 'Bread', 'Butter'],
    ['Bread', 'Butter'],
    ['Milk', 'Bread'],
    ['Milk', 'Eggs'],
    ['Bread', 'Eggs'],
    ['Milk', 'Bread', 'Eggs'],
    ['Milk', 'Bread', 'Butter'],
    ['Bread', 'Butter', 'Eggs'],
    ['Milk', 'Butter'],
    ['Milk', 'Bread', 'Butter', 'Eggs']
]


# ==========================================================
# Step 4: Convert transactions to one-hot encoding
# ==========================================================

te = TransactionEncoder()

te_data = te.fit(transactions).transform(transactions)

df = pd.DataFrame(
    te_data,
    columns=te.columns_
)

print("Transaction Dataset:")
print(df)


# ==========================================================
# Step 5: Find frequent itemsets using FP-Growth
# ==========================================================

frequent_itemsets = fpgrowth(
    df,
    min_support=0.3,
    use_colnames=True
)

print("\n==========================================")
print("FREQUENT ITEMSETS")
print("==========================================")

print(frequent_itemsets)


# ==========================================================
# Step 6: Generate association rules
# ==========================================================

rules = association_rules(
    frequent_itemsets,
    metric="confidence",
    min_threshold=0.5
)


# ==========================================================
# Step 7: Display association rules
# ==========================================================

print("\n==========================================")
print("ASSOCIATION RULES")
print("==========================================")

result = rules[
    [
        'antecedents',
        'consequents',
        'support',
        'confidence',
        'lift'
    ]
]

print(result)


# ==========================================================
# Step 8: Sort rules by lift
# ==========================================================

result_sorted = result.sort_values(
    by='lift',
    ascending=False
)

print("\n==========================================")
print("RULES SORTED BY LIFT")
print("==========================================")

print(result_sorted.to_string(index=False))
```
